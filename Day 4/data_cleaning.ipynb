```python

# Drop columns that are not required for the current business analysis.
# Keep the original dataset unchanged.

columns_to_drop = [
    'search_tags',          # Exclude initially; keywords and search_tag_cat
                            # can support the planned keyword analysis.

    'landing_page',         # Not needed for channel, platform, or creative
                            # performance metrics; useful for landing-page analysis.

    'advertiser_currency',  # Exclude only after confirming USD-normalized costs
                            # are sufficient for the analysis.

    'stats_currency',       # Currency metadata; retain in raw data for validation.

    'currency_code',        # Currency metadata; retain if currency validation is needed.

    'cmi_currency_code',    # Currency metadata; verify its purpose before excluding.

    'exchange_rate',        # Not needed if costs are consistently normalized to USD.

    'timezone',             # Exclude for initial analysis if timestamps are already
                            # standardized to the reporting timezone.

    'network_margin',       # Exclude from the initial scope unless its definition
                            # is relevant to advertising cost or profitability.
]

# Create an analytical copy
df_cleaned = df.drop(columns=columns_to_drop, errors='ignore')

# Check the result
print("Original shape:", df.shape)
print("Cleaned shape:", df_cleaned.shape)
print("Remaining columns:", df_cleaned.columns.tolist())
```

```python

# Check missing values and their percentages
missing_summary = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (df.isnull().mean() * 100).round(2)
})

missing_summary = (
    missing_summary[missing_summary['Missing_Count'] > 0]
    .sort_values('Missing_Percentage', ascending=False)
)

print(missing_summary)
```

```python

# Check whether missing creative metadata occurs on the same records
creative_cols = ['creative_width', 'creative_height', 'template_id']

print(df[creative_cols].isnull().sum())

print(
    "Rows missing all three creative fields:",
    df[creative_cols].isnull().all(axis=1).sum()
)

print(
    "Rows missing approved budget:",
    df['approved_budget'].isnull().sum()
)
```

```python

# Create a copy for advertising performance analysis
df_cleaned = df.drop(columns=['max_bid_cpm'], errors='ignore').copy()

# Keep rows with valid data for the core metrics
df_cleaned = df_cleaned.dropna(
    subset=['campaign_item_id', 'clicks', 'impressions', 'media_cost_usd']
)

# Check the result
print("Original dataset:", df.shape)
print("Cleaned dataset:", df_cleaned.shape)
print("Remaining missing values:")
print(df_cleaned.isnull().sum()[df_cleaned.isnull().sum() > 0])
```

```python

import pandas as pd

# 1. Convert time to datetime
df_cleaned['time'] = pd.to_datetime(
    df_cleaned['time'],
    errors='coerce'
)

# 2. Convert whole-number creative attributes to nullable integers
# Nullable Int64 preserves missing values instead of failing.
integer_columns = [
    'creative_width',
    'creative_height',
    'template_id'
]

for col in integer_columns:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col],
        errors='coerce'
    ).astype('Int64')

# 3. Ensure financial columns are numeric
financial_columns = [
    'approved_budget',
    'campaign_budget_usd',
    'media_cost_usd'
]

for col in financial_columns:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col],
        errors='coerce'
    )

# 4. Verify the updated data types
print(df_cleaned.dtypes)
```

```python

# Parse the original column for diagnosis
parsed = pd.to_datetime(df['time'], errors='coerce')

# Identify values that failed conversion
invalid_mask = parsed.isna() & df['time'].notna()

# Display examples
print("Failed values:")
print(df.loc[invalid_mask, 'time'].astype(str).head(30).to_list())

# Check the original column's most common values
print("\nSample original values:")
print(df['time'].astype(str).sample(
    min(30, len(df)), random_state=42
).to_list())
```

```python

# Parse both date formats safely
date_text = df['time'].astype('string').str.strip()

# First format: YYYY-MM-DD
dates_iso = pd.to_datetime(
    date_text,
    format='%Y-%m-%d',
    errors='coerce'
)

# Second format: DD-MM-YYYY
dates_dmy = pd.to_datetime(
    date_text,
    format='%d-%m-%Y',
    errors='coerce'
)

# Combine both results
df_cleaned['time'] = dates_iso.fillna(dates_dmy)

# Verify the result
print("Missing timestamps:", df_cleaned['time'].isna().sum())
print("Valid timestamps:", df_cleaned['time'].notna().sum())
print(df_cleaned['time'].head(10))
```

```python
# Check completely duplicated rows
print("Exact duplicate rows:", df_cleaned.duplicated().sum())

# Check repeated campaign item IDs
print(
    "Rows with repeated campaign_item_id:",
    df_cleaned.duplicated(subset=['campaign_item_id']).sum()
)

# Inspect repeated IDs
print(
    df_cleaned[
        df_cleaned.duplicated(
            subset=['campaign_item_id'], keep=False
        )
    ].head(10)
)
```

```python
# Values that should not be negative
for col in ['clicks', 'impressions', 'media_cost_usd']:
    print(f"{col} — negative values:", (df_cleaned[col] < 0).sum())

# Clicks should generally not exceed impressions
print(
    "Rows where clicks exceed impressions:",
    (df_cleaned['clicks'] > df_cleaned['impressions']).sum()
)
```
