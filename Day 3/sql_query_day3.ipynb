```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('ecommerce.db')

# Sales dataset
sal = pd.read_csv('sales_cleaned.csv')
sal.to_sql('sales', conn, if_exists='replace', index=False)

# Customers dataset
cust = pd.read_csv('customers_cleaned.csv')
cust.to_sql('customers', conn, if_exists='replace', index=False)

```

```python
columns = [
    'Age_Group',
    'Date_of_Birth',
    'Email',
    'Phone',
    'Pincode',
    'Registration_Date',
    'Customer_Tier'
]

for col in columns:
    conn.execute(f'ALTER TABLE customers DROP COLUMN "{col}"')

conn.commit()
```

```python
pd.read_sql_query("""
SELECT * FROM customers """, conn)
```

```python
columns = [
    'Order_Time',
    'Coupon_Code',
    'Review_Text',
    'Customer_Age',
    'Customer_Age_Group',
    'Rating',
    'City',
    'State',
    'payment_Mode'
]

for col in columns:
    conn.execute(f'ALTER TABLE sales DROP COLUMN "{col}"')

conn.commit()
```

```python
pd.read_sql_query("""
SELECT * FROM sales """, conn)
```

### Which customers have never placed an order?

```python
pd.read_sql_query("""
SELECT c.*
FROM customers c
LEFT JOIN sales s
    ON c.Customer_ID = s.Customer_ID
WHERE s.Customer_ID IS NULL;
""", conn)
```
