**Data Cleaning Insight — Rating Column ⭐**

The `Rating` column contains 129,970 missing values, and all of them coincide with missing `Review_Text`. Further investigation revealed that ratings are available only for delivered orders, while some delivered orders also lack customer feedback.

**Cleaning decision:** Preserved missing ratings as `NULL` instead of imputing artificial values or deleting records. These missing values may represent customers who did not submit feedback, although the underlying reason cannot be confirmed from the data alone.

**Business impact:** Retaining these records preserves valuable sales and operational information while ensuring that customer satisfaction metrics are calculated only from actual ratings.

**Key takeaway:** Not every missing value is a data-quality problem. Sometimes, preserving missingness is the most accurate analytical decision.

```python
cust['Date_of_Birth'] = pd.to_datetime(
    cust['Date_of_Birth'], errors='coerce'
)

cust['Registration_Date'] = pd.to_datetime(
    cust['Registration_Date'], errors='coerce'
)

cust['Pincode'] = cust['Pincode'].astype('string')
cust['Phone'] = cust['Phone'].astype('string')
cust['Customer_ID'] = cust['Customer_ID'].astype('string')
```

```python
for col in ['Gender', 'Age_Group', 'City', 'State', 'Customer_Tier']:
    print(f"\n{col}:")
    print(cust[col].value_counts())
```

```python
for col in ['Gender', 'Age_Group', 'City', 'State', 'Customer_Tier']:
    cust[col] = cust[col].str.strip()
```

```python
print("Age below 18 or above 100:",
      ((cust['Age'] < 18) | (cust['Age'] > 100)).sum())

print("Negative orders:", (cust['Total_Orders'] < 0).sum())

print("Negative spending:", (cust['Total_Spent'] < 0).sum())
```

```python
today = pd.Timestamp.today().normalize()

calculated_age = (
    today.year - cust['Date_of_Birth'].dt.year
    - (
        (today.month < cust['Date_of_Birth'].dt.month)
        | (
            (today.month == cust['Date_of_Birth'].dt.month)
            & (today.day < cust['Date_of_Birth'].dt.day)
        )
    )
)

print("Age mismatches:", (cust['Age'] != calculated_age).sum())
```

```python
print(
    "Registration before birth:",
    (cust['Registration_Date'] < cust['Date_of_Birth']).sum()
)

print(cust.groupby('Customer_Tier').agg(
    Customers=('Customer_ID', 'count'),
    Avg_Orders=('Total_Orders', 'mean'),
    Avg_Spending=('Total_Spent', 'mean')
).round(2))
```
