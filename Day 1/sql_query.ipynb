```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect('sales.db')
```

```python
cursor = conn.cursor()
```

```python
df = pd.read_csv('sales_data_cleaned.csv')

df.to_sql(
    'sales',
    conn,
    if_exists='replace',
    index=False
)
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT * FROM sales
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
product_name,
SUM(units_sold * unit_price) AS total_revenue
FROM sales
GROUP BY product_name
ORDER BY sales_amount DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```
