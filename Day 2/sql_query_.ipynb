```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect('amazon_sales.db')
```

```python
df = pd.read_csv('amazon_sales_cleaned.csv')
```

```python
df.to_sql(
    'amazon_sales',
    conn,
    if_exists='replace',
    index=False
)
```

```python
query = """
SELECT
    "ship-state",
    ROUND(SUM(Amount), 2) AS total_sales
FROM amazon_sales
GROUP BY "ship-state"
ORDER BY total_sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```
