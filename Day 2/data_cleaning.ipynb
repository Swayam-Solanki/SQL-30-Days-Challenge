```python
state_mapping = {
    'New Delhi': 'Delhi',
    'Orissa': 'Odisha',
    'Rj': 'Rajasthan',
    'Rajshthan': 'Rajasthan',
    'Pb': 'Punjab',
    'Punjab/Mohali/Zirakpur': 'Punjab',
    'Nl': 'Nagaland'
}
```

```python
df['ship-state'] = (
    df['ship-state']
    .str.strip()
    .str.title()
)
```

```python
df['ship-state'] = df['ship-state'].replace(state_mapping)
```
